```python
from pypdf import PdfReader

input = 'exemple_cv_2_pages.pdf'

reader = PdfReader('cv_examples/'+input)

print(len(reader.pages))

page = reader.pages[0]

# extracting text from page
text = page.extract_text()
with open("CV_exemple.txt", "w", encoding="utf-8") as fichier:
    fichier.write(text)
```
